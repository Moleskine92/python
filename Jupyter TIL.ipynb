### string interpolation

- %-formatting
- str.format()

````python
num = 10
````

````python
print('페러스 매너스는 %s번 군단의 군주입니다.' %num)
````

## 2.3 논리연산자
- and : 양 쪽 다 True일 때, True를 반환
- or : 양 쪽 모두 False일 때, False를 반환
- not : 상태를 부정

````python
print(True and True)
print(True and False)
print(False and True)
print(False and False)
````

````python
print(True or True)
print(True or False)
print(False or True)
print(False or False)
````

````python
a = True
print(not a)
````

````python
# 단축평가
print(3 and 5)
print(3 and 0)
print(0 and 5)
print(0 and 0)
````
